```python
import scanpy as sc
```

```python
adata = sc.read_h5ad("/Users/coellearth/Desktop/Mammary_Gland_Diet_Project/*originaldata/human_atlas/scRNAseq-human-all-cells.h5ad")
```

```python
adata.obs["broad_cell_type_orig"] = adata.obs["broad_cell_type"].copy()
```

```python
adata.obs["broad_cell_type"] = adata.obs["broad_cell_type"].astype(str)
```

```python
mask_fibro = adata.obs["author_cell_type"].astype(str).str.startswith("Fibro-")
```

```python
adata.obs.loc[mask_fibro, "broad_cell_type"] = (
    adata.obs.loc[mask_fibro, "author_cell_type"]
        .astype(str)
        .str.replace(r"^Fibro", "fibro", regex=True)
)
```

```python
mask_fibro = adata.obs["author_cell_type"].astype(str).str.startswith("Fibro-")
print(adata.obs.loc[mask_fibro, ["author_cell_type", "broad_cell_type"]].head())
```

```python
for col in ["broad_cell_type", "broad_cell_type_orig", "author_cell_type"]:
    if col in adata.obs:
        adata.obs[col] = adata.obs[col].astype("category")
```

```python
from scipy.sparse import issparse
print(type(adata.X), issparse(adata.X))
```

```python
from scipy.sparse import csr_matrix, issparse

if not issparse(adata.X):
    adata.X = csr_matrix(adata.X)
else:
    adata.X = adata.X.tocsr() 
```

```python
print(type(adata.X))
print(adata.X.shape)
print(adata.obs.memory_usage(deep=True).sum())

from scipy.sparse import issparse

if adata.raw is not None:
    print(type(adata.raw.X), issparse(adata.raw.X))

from scipy.sparse import issparse

for k, v in adata.layers.items():
    print(k, type(v), issparse(v))
```

```python
print(type(adata.raw.X), issparse(adata.raw.X))
for k, v in adata.layers.items():
    print(k, type(v), issparse(v))
```
